Importing the libraries for a logistic regression.

```python
#| echo: true
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LogisticRegression
```

Generating the data.

```python
#| echo: true
np.random.seed(42)
X = np.random.normal(0, 2, 100).reshape(-1, 1)
y = (X[:, 0] > 0).astype(int)
flip_idx = np.random.choice(100, size=10, replace=False)
y[flip_idx] = 1 - y[flip_idx]
```

Creating and fitting the model.

```python
#| echo: true
model = LogisticRegression()
model.fit(X, y)
```

Plotting the final result.

```python
#| echo: true
X_test = np.linspace(-6, 6, 300).reshape(-1, 1)
probabilities = model.predict_proba(X_test)[:, 1]

plt.figure(figsize=(8, 5))
plt.scatter(X, y, color='blue', alpha=0.5, label='Observations')
plt.plot(X_test, probabilities, color='red', linewidth=2, label=r'$P(y=1 | x)$')

plt.xlabel("x")
plt.ylabel("P")
plt.legend()
plt.show()
```
