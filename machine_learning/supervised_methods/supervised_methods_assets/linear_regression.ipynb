Importing the libraries for a linear regression.

```python
#| echo: true
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
```

Generating the data.

```python
#| echo: true
X = np.random.uniform(-5, 5, 100).reshape(-1, 1)
y = 2.5 * X.squeeze() + 1.0 + np.random.normal(0, 2, 100)
```

Creating and fitting the model.

```python
#| echo: true
model = LinearRegression()
model.fit(X, y)
```

Printing the coefficients.

```python
#| echo: true
A = model.coef_[0]
b = model.intercept_

print(f"y_hat = {A:.2f} * x + {b:.2f}")
```

Plotting the final result.

```python
#| echo: true
plt.figure(figsize=(8, 5))
plt.scatter(X, y, color='blue', alpha=0.6, label='Observations')
plt.plot(X, model.predict(X), color='red', linewidth=2, label=f'y_hat = {A:.2f}x + {b:.2f}')
plt.xlabel("x")
plt.ylabel("y")
plt.legend()
plt.grid(True, linestyle='--', alpha=0.7)
plt.show()
```
