Importing the packages.

```python
#| echo: true
import torch
import torch.nn as nn
import matplotlib.pyplot as plt
```

Plotting the interpolation between samples from two normal distributions.

```python
#| echo: true
n_samples = 2000
x0 = torch.randn(n_samples, 2) + torch.tensor([-3.0, -3.0])
x1 = torch.randn(n_samples, 2) * 0.5 + torch.tensor([3.0, 3.0])

num_paths = 150
steps = 50

test_x0 = x0[:num_paths]
test_x1 = x1[:num_paths]

t_vals = torch.linspace(0, 1, steps)

trajectories = []
for t in t_vals:
    xt = t * test_x1 + (1 - t) * test_x0
    trajectories.append(xt)

trajectories = torch.stack(trajectories)

plt.figure(figsize=(8, 8))
plt.scatter(x0[:, 0].numpy(), x0[:, 1].numpy(), c='blue', label=r"$p_0$")
plt.scatter(x1[:, 0].numpy(), x1[:, 1].numpy(), c='red', label=r"$p_1$")

for i in range(num_paths):
    plt.plot(trajectories[:, i, 0].numpy(), trajectories[:, i, 1].numpy(), color='black', linewidth=0.8)

plt.legend()
plt.show()
```
